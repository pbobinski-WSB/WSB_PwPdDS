### 1.2. Zmienne i podstawowe typy danych
Zmienna to "pudełko z etykietą", w którym przechowujemy dane do późniejszego wykorzystania.
Główne typy danych w analityce:
- `int` (liczby całkowite) – np. liczba sprzedanych sztuk.
- `float` (liczby zmiennoprzecinkowe) – np. cena, marża, stopy procentowe.
- `str` (tekst / string) – np. nazwa klienta, kategoria produktu.
- `bool` (wartość logiczna) – tylko `True` lub `False` (np. czy faktura opłacona).

```python
nazwa_produktu = "Laptop AI Pro"   # str
liczba_sztuk = 5                    # int
cena_jednostkowa = 4500.50          # float
czy_dostepny = True                 # bool

# Sprawdzanie typu zmiennej za pomocą type()
print(type(nazwa_produktu))
print(type(cena_jednostkowa))

# Nowoczesne formatowanie tekstu: f-string
wartosc_zamowienia = liczba_sztuk * cena_jednostkowa
print(f"Produkt: {nazwa_produktu} | Łączna wartość: {wartosc_zamowienia} PLN")
```

### 1.3. Konwersja typów (Casting)
Często dane pobrane z pliku lub bazy przychodzą jako tekst (`str`), nawet jeśli wyglądają jak liczby.
Musimy je jawnie przekonwertować: `int()`, `float()`, `str()`.

```python
cena_tekst = "100"
liczba_tekst = "5"

# Błąd logiczny: zamiast 100 * 5, dostaniemy powtórzony tekst lub błąd
# print(cena_tekst * liczba_tekst) -> BŁĄD!

# Poprawna konwersja:
cena_liczba = float(cena_tekst)
liczba_sztuk = int(liczba_tekst)
print(f"Prawidłowy wynik: {cena_liczba * liczba_sztuk} PLN")
```
