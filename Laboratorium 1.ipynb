### 1.2. Zmienne i podstawowe typy danych
Zmienna to "pudełko z etykietą", w którym przechowujemy dane do późniejszego wykorzystania.
Główne typy danych w analityce:
- `int` (liczby całkowite) – np. liczba sprzedanych sztuk.
- `float` (liczby zmiennoprzecinkowe) – np. cena, marża, stopy procentowe.
- `str` (tekst / string) – np. nazwa klienta, kategoria produktu.
- `bool` (wartość logiczna) – tylko `True` lub `False` (np. czy faktura opłacona).

```python
nazwa_produktu = "Laptop AI Pro"   # str
liczba_sztuk = 5                    # int
cena_jednostkowa = 4500.50          # float
czy_dostepny = True                 # bool

# Sprawdzanie typu zmiennej za pomocą type()
print(type(nazwa_produktu))
print(type(cena_jednostkowa))

# Nowoczesne formatowanie tekstu: f-string
wartosc_zamowienia = liczba_sztuk * cena_jednostkowa
print(f"Produkt: {nazwa_produktu} | Łączna wartość: {wartosc_zamowienia} PLN")
```

### 1.3. Konwersja typów (Casting)
Często dane pobrane z pliku lub bazy przychodzą jako tekst (`str`), nawet jeśli wyglądają jak liczby.
Musimy je jawnie przekonwertować: `int()`, `float()`, `str()`.

```python
cena_tekst = "100"
liczba_tekst = "5"

# Błąd logiczny: zamiast 100 * 5, dostaniemy powtórzony tekst lub błąd
# print(cena_tekst * liczba_tekst) -> BŁĄD!

# Poprawna konwersja:
cena_liczba = float(cena_tekst)
liczba_sztuk = int(liczba_tekst)
print(f"Prawidłowy wynik: {cena_liczba * liczba_sztuk} PLN")
```

### 2.2. Instrukcja `if ... elif ... else`
**UWAGA NA WCIĘCIA!** W Pythonie bloki kodu definiujemy za pomocą wcięć (standardowo 4 spacje lub 1 Tab).

```python
# [Scoring rabatowy]
wartosc_koszyka = 1200.0
czy_klient_vip = True

if wartosc_koszyka > 1000 and czy_klient_vip:
    rabat = 0.20  # 20% rabatu
    print("Przyznano maksymalny rabat VIP: 20%")
elif wartosc_koszyka > 1000 or czy_klient_vip:
    rabat = 0.10  # 10% rabatu
    print("Przyznano standardowy rabat: 10%")
else:
    rabat = 0.0
    print("Brak rabatu.")

kwota_po_rabacie = wartosc_koszyka * (1 - rabat)
print(f"Kwota do zapłaty: {kwota_po_rabacie} PLN")
```

### 3.1. Listy (`list`) – uporządkowane serie danych
Lista przechowuje wiele elementów w jednej zmiennej.
- Indeksowanie zaczyna się od **0**!
- Ostatni element to indeks `-1`.
- Dodawanie nowego elementu: metoda `.append()`.

```python
# [PRZYKŁAD]
ceny = [19.99, 49.99, 99.00, 150.50]

print(f"Pierwsza cena: {ceny[0]}")
print(f"Ostatnia cena: {ceny[-1]}")
print(f"Liczba produktów: {len(ceny)}")

# Dodanie nowego produktu na koniec listy
ceny.append(299.99)
print(f"Lista po aktualizacji: {ceny}")

# Podstawowe statystyki wbudowane w Pythona
print(f"Najniższa cena: {min(ceny)} PLN")
print(f"Najwyższa cena: {max(ceny)} PLN")
print(f"Suma cen: {sum(ceny)} PLN")
```

### 4.1. Pętla `for` – automatyczne przetwarzanie kolekcji
Pętla pozwala wykonać tę samą operację dla każdego elementu na liście.

```python
# [PRZYKŁAD: Przeglądanie listy transakcji]
transakcje = [120, 450, 890, 25, 1300, 75]

print("Transakcje o wysokiej wartości (> 200 PLN):")
suma_wysokich = 0

for kwota in transakcje:
    if kwota > 200:
        print(f"-> Znaleziono: {kwota} PLN")
        suma_wysokich += kwota

print(f"Łączna suma dużych transakcji: {suma_wysokich} PLN")
```

Masz listę miesięcznych przychodów pewnego startupu AI z całego roku:
`przychody = [12000, 15000, 9000, 18000, 22000, 25000, 19000, 31000, 28000, 35000, 42000, 50000]`

Napisz pętlę `for`, która:
1. Zliczy, ile miesięcy miało przychód większy lub równy **25 000 PLN**.
2. Obliczy średni przychód w całym roku (wskazówka: użyj `sum()` i `len()`).
3. Wyświetli raport z wynikami.

```python
# [ROZWIĄZANIE ZADANIA 4]
przychody = [12000, 15000, 9000, 18000, 22000, 25000, 19000, 31000, 28000, 35000, 42000, 50000]

dobre_miesiace = 0
for p in przychody:
    if p >= 25000:
        dobre_miesiace += 1

srednia = sum(przychody) / len(przychody)

print(f"Liczba miesięcy z przychodem >= 25k: {dobre_miesiace}")
print(f"Średni miesięczny przychód: {srednia:.2f} PLN")
```

### 5.1. Tworzenie funkcji (`def`)
Funkcja to wydzielony blok kodu, który:
1. Przyjmuje argumenty wejściowe.
2. Przetwarza je.
3. Zwraca wynik słowem kluczowym `return`.

```python
# [PRZYKŁAD]
def oblicz_cene_brutto(cena_netto, stawka_vat=0.23):
    """Oblicza cenę brutto na podstawie ceny netto i stawki VAT."""
    cena_brutto = cena_netto * (1 + stawka_vat)
    return round(cena_brutto, 2)

# Użycie funkcji
produkt_a = oblicz_cene_brutto(100.0)       # domyślny VAT 23%
produkt_b = oblicz_cene_brutto(100.0, 0.08) # książka - VAT 8%

print(f"Produkt A brutto: {produkt_a} PLN")
print(f"Produkt B brutto: {produkt_b} PLN")
```

### 🏆 FINAŁOWY PROJEKT ZAJĘĆ (Case Study): Analizator Rentowności Produktów
Napisz kompletny mini-program analityczny dla sklepu internetowego.

**Krok 1:** Napisz funkcję `analizuj_produkt(nazwa, cena_sprzedazy, koszt_magazynowy)`, która:
- Obliczy marżę kwotową: `marza = cena_sprzedazy - koszt_magazynowy`
- Obliczy marżę procentową (narzut): `marza_proc = (marza / koszt_magazynowy) * 100`
- Zwróci status:
  - `"BARDZO ZYSKOWNY"` jeśli `marza_proc >= 50%`
  - `"UMIARKOWANY"` jeśli `marza_proc` między `20%` a `50%`
  - `"NIERENTOWNY"` jeśli `marza_proc < 20%`
- Funkcja powinna zwrócić słownik z podsumowaniem: `{"nazwa": nazwa, "marza": marza, "status": status}`

**Krok 2:** Wykorzystaj funkcję w pętli dla poniższej listy produktów z magazynu i wypisz raport:

```python
# Dane wejściowe do projektu:
produkty_magazyn = [
    {"nazwa": "Słuchawki Bezprzewodowe", "cena": 199.0, "koszt": 80.0},
    {"nazwa": "Kabel USB-C", "cena": 29.0, "koszt": 25.0},
    {"nazwa": "Smartwatch AI", "cena": 899.0, "koszt": 500.0},
    {"nazwa": "Podkładka pod mysz", "cena": 39.0, "koszt": 35.0}
]
```

```python
# [ROZWIĄZANIE PROJEKTU FINAŁOWEGO]
def analizuj_produkt(nazwa, cena_sprzedazy, koszt_magazynowy):
    marza = cena_sprzedazy - koszt_magazynowy
    marza_proc = (marza / koszt_magazynowy) * 100
    
    if marza_proc >= 50:
        status = "BARDZO ZYSKOWNY"
    elif marza_proc >= 20:
        status = "UMIARKOWANY"
    else:
        status = "NIERENTOWNY"
        
    return {
        "nazwa": nazwa,
        "marza": round(marza, 2),
        "marza_proc": round(marza_proc, 1),
        "status": status
    }

print("=== RAPORT RENTOWNOŚCI MAGAZYNU ===")
for p in produkty_magazyn:
    wynik = analizuj_produkt(p["nazwa"], p["cena"], p["koszt"])
    print(f"Produkt: {wynik['nazwa']:<25} | Marża: {wynik['marza']:>6.2f} zł ({wynik['marza_proc']}%) | Status: {wynik['status']}")
```
